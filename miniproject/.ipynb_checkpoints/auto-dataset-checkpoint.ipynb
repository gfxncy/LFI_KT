```python
import vtk

# Создаем reader для VTK файла
reader = vtk.vtkGenericDataObjectReader()
reader.SetFileName("data/potential000000.vtu)
reader.Update()

# Получаем данные
data = reader.GetOutput()

# Если это структурированные точки (StructuredPoints/ImageData)
if data.IsA("vtkStructuredPoints") or data.IsA("vtkImageData"):
    print("Dimensions:", data.GetDimensions())
    print("Spacing:", data.GetSpacing())
    print("Origin:", data.GetOrigin())
    print("Scalar range:", data.GetScalarRange())

# Если это полигональные данные (vtkPolyData)
elif data.IsA("vtkPolyData"):
    print("Number of points:", data.GetNumberOfPoints())
    print("Number of cells:", data.GetNumberOfCells())
```
